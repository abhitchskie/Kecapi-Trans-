```python
import pandas as pd

# 1. Kita bikin data mentah (kayak struct/array di C)
data_pertandingan = {
    'Tim_Kandang': ['Arsenal', 'Man City', 'Liverpool', 'Chelsea'],
    'Tim_Tandang': ['Spurs', 'MU', 'Everton', 'Fulham'],
    'Gol_Kandang': [3, 2, 1, 0],
    'Gol_Tandang': [1, 1, 0, 0],
    'xG_Kandang': [2.4, 1.8, 0.9, 0.4], # Expected Goals (Peluang)
    'xG_Tandang': [0.5, 1.2, 0.2, 0.3]
}

# 2. Mengubah data mentah jadi tabel pintar bernama DataFrame
df = pd.DataFrame(data_pertandingan)

# 3. Menambahkan logika sederhana: Siapa yang menang?
# Kalau Gol Kandang > Gol Tandang = Home Win (H), dst.
import numpy as np
kondisi = [
    df['Gol_Kandang'] > df['Gol_Tandang'],
    df['Gol_Kandang'] < df['Gol_Tandang']
]
pilihan = ['Home Win', 'Away Win']
df['Hasil_Akhir'] = np.select(kondisi, pilihan, default='Draw')

# Tampilkan hasilnya
df
```
